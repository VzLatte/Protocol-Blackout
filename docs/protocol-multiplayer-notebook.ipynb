
## 2) Code Changes Summary
- Add `hasSubmitted` to `Player` type and mapper.
- Add `WaitingForOpponent` UI component.
- Show waiting overlay in `TurnEntryView` after submit (includes 15s countdown and auto-submit on timeout).
- Use a 15s resolution countdown in `ResolutionView` and show "Awaiting server resolution..." when logs are not yet available.
- Ensure `ResolutionView` uses a fallback snapshot of players so avatars are visible during resolution.

## 4) Manual verification steps
1. Start server and client:
   - `cd protocol-server && npm install && npm run dev`
   - `cd protocol-client && npm install && npm run dev`
2. Start two browser windows (or two private windows) and join the same multiplayer room.
3. In Turn Entry stage, submit from Player A and verify Player A sees a "Waiting for opponents..." overlay with countdown.
4. Player B submits — both clients should transition to Resolution and see the 15s resolution countdown.
5. Verify the resolution log shows either actions or, if none yet, displays "Awaiting server resolution..." with spinner.
6. Verify players are visible on the map during resolution (avatars or fallback icons).
